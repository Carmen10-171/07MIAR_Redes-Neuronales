#### **- Batch Normalization**

````python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, BatchNormalization, Dropout

# Definimos la variable que faltaba
WORD_LIMIT = 10000

# Vamos a codificar la topología de nuestro MLP
model_bn = Sequential()
model_bn.add(Dense(128,activation='relu', input_shape=(WORD_LIMIT,)))
model_bn.add(BatchNormalization())
model_bn.add(Dropout(0.75))
model_bn.add(Dense(64,activation='relu'))
model_bn.add(BatchNormalization())
model_bn.add(Dropout(0.75))
model_bn.add(Dense(46,activation='softmax')) # Reparto de la unidad de probabilidad entre num_classes
````

````python
# Ahora que tengo definida la arquitectura, la compilo con el optimizador Adam
model_bn.compile(
    optimizer='adam',
    loss='categorical_crossentropy',
    metrics=['accuracy']
)
````

````python
import numpy as np
import tensorflow as tf
from tensorflow.keras.utils import to_categorical

# Definimos de forma segura la función de codificación
def one_hot_encode(data):
    encoded = np.zeros((len(data), WORD_LIMIT))
    for i, v in enumerate(data):
        encoded[i, v] = 1.0
    return encoded

# Nos aseguramos de cargar y codificar los datos de Reuters
reuters = tf.keras.datasets.reuters
(training_data, training_labels), (testing_data, testing_labels) = reuters.load_data(num_words=WORD_LIMIT)

x_train_reuters = one_hot_encode(training_data)
y_train_reuters = to_categorical(training_labels)

# Entrenamos el modelo model_bn
H = model_bn.fit(
    x_train_reuters,
    y_train_reuters,
    epochs=20,
    batch_size=32,
    validation_split=0.2
)
````

````python
import matplotlib.pyplot as plt
# Muestro gráfica de accuracy y losses
plt.style.use("ggplot")
plt.figure()
plt.plot(np.arange(0, 20), H.history["loss"], label="train_loss")
plt.plot(np.arange(0, 20), H.history["val_loss"], label="val_loss")
plt.plot(np.arange(0, 20), H.history["accuracy"], label="train_acc")
plt.plot(np.arange(0, 20), H.history["val_accuracy"], label="val_acc")
plt.title("Training Loss and Accuracy (Batch Normalization & Dropout)")
plt.xlabel("Epoch #")
plt.ylabel("Loss/Accuracy")
plt.legend()
plt.show()
````

````python
# Codificamos de manera segura el conjunto de test de Reuters
x_test_reuters = one_hot_encode(testing_data)
y_test_reuters = to_categorical(testing_labels)

# Evaluando el modelo de predicción con el subset de test de Reuters
print("[INFO]: Evaluando red neuronal...")
model_bn.predict(x_test_reuters)
loss, accuracy = model_bn.evaluate(x_test_reuters, y_test_reuters)
print('Loss {}, accuracy {}'.format(loss, accuracy))
````

````python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Dropout

# 1. Definimos la arquitectura idéntica pero SIN Batch Normalization
model_no_bn = Sequential([
    Dense(128, activation='relu', input_shape=(WORD_LIMIT,)),
    Dropout(0.75),
    Dense(64, activation='relu'),
    Dropout(0.75),
    Dense(46, activation='softmax')
])

# 2. Compilamos el modelo con los mismos parámetros
model_no_bn.compile(
    optimizer='adam',
    loss='categorical_crossentropy',
    metrics=['accuracy']
)

# 3. Entrenamos el modelo sin BN
print("[INFO]: Entrenando modelo de control (Sin Batch Normalization)...")
H_no_bn = model_no_bn.fit(
    x_train_reuters,
    y_train_reuters,
    epochs=20,
    batch_size=32,
    validation_split=0.2,
    verbose=1
)

# 4. Evaluamos en el conjunto de test
print("\n[INFO]: Evaluando modelo de control...")
loss_no_bn, acc_no_bn = model_no_bn.evaluate(x_test_reuters, y_test_reuters, verbose=0)

# 5. Comparamos los resultados
print("\n================ COMPARATIVA DE RENDIMIENTO ================")
print("Modelo CON Batch Normalization:")
print(f"  - Loss en Test: {loss:.4f}")
print(f"  - Accuracy en Test: {accuracy:.4f}")
print("\nModelo SIN Batch Normalization:")
print(f"  - Loss en Test: {loss_no_bn:.4f}")
print(f"  - Accuracy en Test: {acc_no_bn:.4f}")
print("===========================================================")
````
